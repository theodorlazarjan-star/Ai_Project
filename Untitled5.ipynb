This is a pipeline. With this code we can train several classification models and predict our churn and evaluate it on different scores.
This allows us to compare the models and see which one performs best for predicting churn.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=1000, random_state=42))
    ]),

    "kNN": Pipeline([
        ("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=5))
    ]),

    "Naive Bayes": Pipeline([
        ("scaler", StandardScaler()),
        ("model", GaussianNB())
    ]),

    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(probability=True, random_state=42))
    ]),

    "Decision Tree": Pipeline([
        ("model", DecisionTreeClassifier(random_state=42))
    ]),

    "Random Forest": Pipeline([
    ("model", RandomForestClassifier(random_state=42))
])

}

for name, pipeline in models.items():

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    print(f"\n{name}")
    print("Accuracy :", accuracy_score(y_test, y_pred))
    print("Precision:", precision_score(y_test, y_pred))
    print("Recall   :", recall_score(y_test, y_pred))
    print("F1-score :", f1_score(y_test, y_pred))

    cm = confusion_matrix(y_test, y_pred)
    ConfusionMatrixDisplay(cm, display_labels=["No Churn", "Churn"]).plot(cmap="Blues")
    plt.title(name)
    plt.show()
```

This code uses 5-fold cross-validation to evaluate each model multiple times on different parts of the training data. It calculates the F1-score for each fold, the mean F1-score, and the standard deviation to check how well and how consistently each model performs.

```python
from sklearn.model_selection import cross_val_score

cv_results = {}

for name, pipeline in models.items():

    scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=5,
        scoring="f1",
        n_jobs=-1
    )

    cv_results[name] = scores

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print("Fold Scores:", scores)
    print("Mean F1:", round(scores.mean(), 4))
    print("Std:", round(scores.std(), 4))
```

Here we tune our hyperparameters to find the best parameters for our model

```python
param_grids = {
    "Logistic Regression": {
        "model__C": [0.1, 1, 10]
    },
    "kNN": {
        "model__n_neighbors": [5, 9, 13]
    },
    "Naive Bayes": {
        "model__var_smoothing": [1e-9, 1e-8]
    },
    "SVM": {
        "model__C": [0.1, 1, 10],
        "model__kernel": ["linear", "rbf"]
    },
    "Decision Tree": {
        "model__max_depth": [5, 10, None],
        "model__min_samples_split": [2, 5]
    },
    "Random Forest": {
        "model__n_estimators": [100, 200],
        "model__max_depth": [5, 10],
        "model__min_samples_split": [2, 5]
    }
}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

best_models = {}

for name, pipeline in models.items():

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    grid = GridSearchCV(
        pipeline,
        param_grids[name],
        cv=5,
        scoring="f1",
        n_jobs=-1
    )

    grid.fit(X_train, y_train)

    best_models[name] = grid.best_estimator_

    print("Best Parameters:", grid.best_params_)
    print("Best CV F1:", round(grid.best_score_, 4))

    y_pred = grid.predict(X_test)

    print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
    print("Precision:", round(precision_score(y_test, y_pred), 4))
    print("Recall   :", round(recall_score(y_test, y_pred), 4))
    print("F1-score :", round(f1_score(y_test, y_pred), 4))
```

Here we trie to optimite the threshold for the models where we can change them

```python
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

thresholds = np.arange(0.1, 1.0, 0.05)

threshold_results = {}

for name, model in best_models.items():

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    y_prob = model.predict_proba(X_test)[:, 1]

    best_accuracy = (0, 0)
    best_precision = (0, 0)
    best_recall = (0, 0)
    best_f1 = (0, 0)

    for threshold in thresholds:

        y_pred = (y_prob >= threshold).astype(int)

        accuracy = accuracy_score(y_test, y_pred)

        precision = precision_score(
            y_test,
            y_pred,
            zero_division=0
        )

        recall = recall_score(
            y_test,
            y_pred,
            zero_division=0
        )

        f1 = f1_score(
            y_test,
            y_pred,
            zero_division=0
        )

        if accuracy > best_accuracy[1]:
            best_accuracy = (threshold, accuracy)

        if precision > best_precision[1]:
            best_precision = (threshold, precision)

        if recall > best_recall[1]:
            best_recall = (threshold, recall)

        if f1 > best_f1[1]:
            best_f1 = (threshold, f1)

    threshold_results[name] = {
        "Accuracy": best_accuracy,
        "Precision": best_precision,
        "Recall": best_recall,
        "F1": best_f1
    }

    print("Best Accuracy :", best_accuracy)
    print("Best Precision:", best_precision)
    print("Best Recall   :", best_recall)
    print("Best F1       :", best_f1)
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1"]

for metric in metrics:

    best_model_name = max(
        threshold_results,
        key=lambda name: threshold_results[name][metric][1]
    )

    best_score = threshold_results[best_model_name][metric][1]
    best_threshold = threshold_results[best_model_name][metric][0]
    best_parameters = best_models[best_model_name].get_params()

    print("\n" + "=" * 50)
    print(f"BEST MODEL FOR {metric.upper()}")
    print("=" * 50)

    print("Model:", best_model_name)
    print("Score:", round(best_score, 4))
    print("Threshold:", best_threshold)

    print("Best Hyperparameters:")

    for parameter, value in best_parameters.items():
        if parameter.startswith("model__"):
            print(f"  {parameter}: {value}")
```

After we have finished training our models, we can look at the feature importance from our Logistic Regression, Decision Tree, and Random Forest models. It shows how important each feature was for the model's predictions. With this, we can identify irrelevant or less useful features and potentially remove them.

```python
import pandas as pd
import matplotlib.pyplot as plt

model = best_models["Logistic Regression"].named_steps["model"]
importance = pd.Series(model.coef_[0], index=X.columns).sort_values()
plt.figure(figsize=(10,6))
importance.plot(kind="barh")
plt.title("Logistic Regression Feature Coefficients")
plt.xlabel("Coefficient")
plt.ylabel("Features")
plt.axvline(0, color="black", linewidth=1)
plt.show()

model = best_models["Decision Tree"].named_steps["model"]
importance = pd.Series(model.feature_importances_, index=X.columns).sort_values()
plt.figure(figsize=(10,6))
importance.plot(kind="barh")
plt.title("Decision Tree Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Features")
plt.show()

model = best_models["Random Forest"].named_steps["model"]
importance = pd.Series(model.feature_importances_, index=X.columns).sort_values()
plt.figure(figsize=(10,6))
importance.plot(kind="barh")
plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Features")
plt.show()
```

In the end, I chose the F1-score because using recall alone is not that good. A model could achieve a perfect recall by predicting that every customer will churn. The F1-score balances precision and recall, making it a better metric for this churn prediction problem.

Here we set our final model and threshold

```python
final_model = best_models["Random Forest"]
final_threshold = 0.30
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

X_final_test = df_final_test.drop(columns=["Churn"])
y_final_test = df_final_test["Churn"]

X_final_test = X_final_test[final_model.feature_names_in_]

y_final_prob = final_model.predict_proba(X_final_test)[:, 1]

y_final_pred = (y_final_prob >= final_threshold).astype(int)

final_accuracy = accuracy_score(y_final_test, y_final_pred)
final_precision = precision_score(y_final_test, y_final_pred, zero_division=0)
final_recall = recall_score(y_final_test, y_final_pred, zero_division=0)
final_f1 = f1_score(y_final_test, y_final_pred, zero_division=0)

print("\n" + "=" * 50)
print("FINAL TEST RESULTS")
print("=" * 50)

print("Threshold:", final_threshold)

print("Accuracy :", round(final_accuracy, 4))
print("Precision:", round(final_precision, 4))
print("Recall   :", round(final_recall, 4))
print("F1-score :", round(final_f1, 4))
```
